# Config | SDMA placement
Run the two code cells in order. The first selects the inputs; **Run SDMA** compiles the circuits and writes a layout PDF and an execution notebook. This notebook runs locally and needs no IBM credentials.

## 1. Inputs and settings
The calibration is an IBM export. The policy CSV gives the depth bin and minimum horizontal and vertical distances. The workload CSV lists `path,times`, with circuit paths relative to `CIRCUIT_DIR`. The supplied example contains 80 draws with replacement. Edit the three paths to use your own CSVs.

```python
from pathlib import Path

ROOT = Path.cwd()  # If your kernel starts elsewhere, set this to the folder containing Config.ipynb.
EXAMPLE = ROOT / "Example"
CALIBRATION_CSV = EXAMPLE / "ibm_kingston_calibrations_2026-11-13T14_57_10Z.csv"
POLICY_CSV = EXAMPLE / "policy.csv"
WORKLOAD_CSV = EXAMPLE / "workload.csv"
CIRCUIT_DIR = ROOT / "circuits"
OUTPUT_DIR = EXAMPLE / "output"
TOPOLOGY = ROOT / "sdma" / "heron_r2.json"

SEPARATE_VALUE = 0.5           # Depth-ratio rounding threshold s, from 0 to 1.
OPTIMIZATION_LEVEL = 2         # Qiskit levels: 0, 1, 2, or 3.
ROUTING_METHOD = "sabre"
TRANSPILER_SEED = 20260928
PLACEMENT_SEED = 42

# Optional IBM execution settings. None leaves the Runtime default in effect.
INIT_QUBITS = None
REP_DELAY_SECONDS = None
MAX_EXECUTION_TIME_SECONDS = None
```

## 2. Run SDMA
Run this cell after changing any settings. Open the PDF to inspect the layout, then open **Execute.ipynb** to submit pages when ready. When the output directory already contains a run, the cell links to a new timestamped subfolder so earlier jobs and results remain accessible.

```python
import sys
from IPython.display import Markdown, display
from urllib.parse import quote
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
from place import allocate

run_dir = allocate(
    calibration=CALIBRATION_CSV, policy=POLICY_CSV, workload=WORKLOAD_CSV,
    circuits=CIRCUIT_DIR, output=OUTPUT_DIR, topology=TOPOLOGY, s=SEPARATE_VALUE,
    optimization_level=OPTIMIZATION_LEVEL, routing_method=ROUTING_METHOD,
    seed_transpiler=TRANSPILER_SEED, seed=PLACEMENT_SEED,
    init_qubits=INIT_QUBITS, rep_delay_seconds=REP_DELAY_SECONDS,
    max_execution_time_seconds=MAX_EXECUTION_TIME_SECONDS,
)
if run_dir.is_relative_to(ROOT):
    run_link = quote(run_dir.relative_to(ROOT).as_posix())
    display(Markdown(f"[Layout PDF]({run_link}/layout.pdf) | [Open Execute.ipynb]({run_link}/Execute.ipynb)"))
```
